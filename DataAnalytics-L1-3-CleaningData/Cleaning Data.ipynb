## 3. Data Quality Report

Before performing any cleaning operations, the dataset was assessed to identify potential data quality issues. The assessment covers missing values, duplicate records, data types, categorical consistency, and numerical value/range anomalies.

The findings from this assessment will be used to determine the appropriate cleaning strategy for each issue.


### 3.1 Dataset Overview

### 3.2 Duplicate Records

-The dataset was checked completely  for duplicate rows.but, No duplicate rows were found, with a total of **0 duplicate records**.

-Therefore, no rows need to be removed due to exact duplication.

### 3.3 Missing Vlaues

-Missing-value findings: Age has 177 missing values (19.87%), Cabin has 687 missing values (77.10%), and Embarked has 2 missing values (0.22%). The remaining columns contain no missing values...

### 3.4 Categorical-value findings

-Categorical-value findings: The categorical columns contain consistent and valid values. No unexpected categories or inconsistent formatting were identified, so no changes were required.

### 3.5 Current finding

-Numerical-value findings: Age, SibSp, and Parch do not show obvious invalid values. Fare contains some zero values, which require further investigation before deciding whether they should be treated as errors or valid observations.

-Fare anomaly findings: Fifteen passengers have a fare of 0. These passengers were investigated using their ticket, SibSp, and Parch values. Several passengers share the same ticket numbers, while others have unique tickets. Since there is insufficient evidence to confirm that the zero fares are data-entry errors, the values were retained as valid observations.

### 3.6 Data Types
Data-type findings: The dataset contains appropriate data types overall. Integer columns with small ranges were downcast to smaller integer types to improve memory efficiency. No changes were required for categorical or floating-point columns.

### 3.7 Data Types
Data-type findings: Numeric columns were assigned appropriate integer or floating-point types based on their value ranges and meaning. Text-based columns were converted to the string dtype. Fare and Age were retained as floating-point values because they may contain decimal values

###3.8 Outlier Detection and Treatment

The IQR method was used to identify potential outliers in the numerical columns. The analysis identified 66 outliers in `Age`, 46 in `SibSp`, 213 in `Parch`, and 116 in `Fare`.

-These observations were inspected and retained because the detected values can represent legitimate passengers or travel conditions rather than data-entry errors. For example, higher values of `SibSp` and `Parch` can represent passengers traveling with larger family groups, while higher `Fare` values can be legitimate fares for higher-class passengers.

--so, **no outliers were removed or capped**.

###3.9 Before vs After Cleaning Summary

The dataset initially contained 891 rows and 12 columns. It had 866 missing values across `Age`, `Cabin`, and `Embarked`. No duplicate rows were identified.

After cleaning, all missing values were handled, the `Cabin` column was removed due to its high proportion of missing values, and no rows were removed because there were no duplicate records or confirmed invalid outliers. The cleaned dataset contains 891 rows and 11 columns with corrected data types.

```python
df['Survived'] = df['Survived'].astype('int8')
df['Pclass'] = df['Pclass'].astype('int8')
df['SibSp'] = df['SibSp'].astype('int8')
df['Parch'] = df['Parch'].astype('int8')

df['Name'] = df['Name'].astype('string')
df['Sex'] = df['Sex'].astype('string')
df['Ticket'] = df['Ticket'].astype('string')
df['Embarked'] = df['Embarked'].astype('string')
```

```python
df.describe()
```

```python
numeric_cols = ['Age', 'SibSp', 'Parch', 'Fare']

Q1 = df[numeric_cols].quantile(0.25)
Q3 = df[numeric_cols].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outlier_count = (
    (df[numeric_cols] < lower) |
    (df[numeric_cols] > upper)
).sum()

outlier_count
```
